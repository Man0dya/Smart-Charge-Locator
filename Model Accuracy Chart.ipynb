```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Create a DataFrame summarizing the metrics
data = {
    'Model': ['Linear Regression', 'Ridge Regression', 'Random Forest', 'XGBoost'],
    'R2_Train': [1.0000, 1.0000, 0.7780, 0.9995],
    'R2_Test': [1.0000, 1.0000, 0.9742, 0.9634],
    'MSE_Train': [0.0009, 2.6028, 229872.95, 494.64],
    'MSE_Test': [0.0014, 0.7552, 8747.05, 12417.09],
    'MAE_Train': [0.0176, 0.5702, 43.1350, 3.4932],
    'MAE_Test': [0.0180, 0.6158, 20.5873, 27.8411]
}

df = pd.DataFrame(data)

# Set plot style
sns.set(style="whitegrid")
plt.figure(figsize=(14, 8))

# Plot R² scores
plt.subplot(1, 3, 1)
sns.barplot(x='Model', y='R2_Train', data=df, color='skyblue', label='Train')
sns.barplot(x='Model', y='R2_Test', data=df, color='salmon', label='Test', alpha=0.7)
plt.title('R² Comparison')
plt.xticks(rotation=45)
plt.ylabel('R²')
plt.legend()

# Plot MSE
plt.subplot(1, 3, 2)
sns.barplot(x='Model', y='MSE_Train', data=df, color='skyblue', label='Train')
sns.barplot(x='Model', y='MSE_Test', data=df, color='salmon', label='Test', alpha=0.7)
plt.title('MSE Comparison')
plt.xticks(rotation=45)
plt.ylabel('MSE')
plt.yscale('log')  # Log scale for better visualization

# Plot MAE
plt.subplot(1, 3, 3)
sns.barplot(x='Model', y='MAE_Train', data=df, color='skyblue', label='Train')
sns.barplot(x='Model', y='MAE_Test', data=df, color='salmon', label='Test', alpha=0.7)
plt.title('MAE Comparison')
plt.xticks(rotation=45)
plt.ylabel('MAE')

plt.tight_layout()
plt.show()
```
